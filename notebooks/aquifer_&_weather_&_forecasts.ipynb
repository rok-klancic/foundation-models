# Aquifer data/Surface data + Weather data + Weather forecasts

## Imports

```python
import joblib
import pandas as pd
import matplotlib.pyplot as plt
```

## Aquifer data

### Merge the data

```python
aquifer_by_stations = joblib.load('../data/interim/ground-water-and-weather.joblib')
aquifer_by_stations[85065]
```

```python
weather_forecast = joblib.load('../data/interim/weather-forecast-slovenia-5-days.joblib')

weather_forecast['date'] = weather_forecast.index

weather_forecast
```

```python
for key, station in aquifer_by_stations.items():
    aquifer_by_stations[key] = pd.merge(station, weather_forecast, on='date', how='left')
```

```python
aquifer_by_stations[85065]
```

### Save the data

```python
joblib.dump(aquifer_by_stations, '../data/interim/ground-water-and-weather-with-forecasts-and-additional-features.joblib')
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import minmax_scale
plt.plot(data[85065]['date'][-95:], minmax_scale(data[85065]['precipitation_intensity_avg'][-95:]), label='percipitation_probability')
plt.plot(data[85065]['date'][-95:], minmax_scale(data[85065]['rr_decodeText_5'][-100:-5]), label='rr_decodeText_0')
plt.legend()
plt.show()
```
