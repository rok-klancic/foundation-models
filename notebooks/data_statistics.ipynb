## Aquifer data

```python
import joblib
```

```python
aquifer_by_stations = joblib.load('../data/interim/ground-water-and-weather-with-forecasts-and-additional-features.joblib')
```

```python
list(aquifer_by_stations.keys())
```

```python
aquifer_by_stations[1005].head()
```

```python
list(aquifer_by_stations[1005].columns)[-30:]
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Basic statistics for each station
print(f"Total number of stations: {len(aquifer_by_stations)}")

# Create a summary dataframe
summary_stats = []

for station_id, df in aquifer_by_stations.items():
    stats = {
        'station_id': station_id,
        'num_records': len(df),
        'start_date': df['date'].min(),
        'end_date': df['date'].max(),
        'date_range_days': (df['date'].max() - df['date'].min()).days,
        'missing_values_pct': df.isna().mean().mean() * 100,
        'altitude_diff_mean': df['altitude_diff'].mean() if 'altitude_diff' in df.columns else np.nan,
        'altitude_diff_std': df['altitude_diff'].std() if 'altitude_diff' in df.columns else np.nan
    }
    summary_stats.append(stats)

summary_df = pd.DataFrame(summary_stats)
display(summary_df)


# Get a sample station to examine its columns
sample_station_id = list(aquifer_by_stations.keys())[0]
sample_df = aquifer_by_stations[sample_station_id]
print(f"\nColumns in sample station {sample_station_id}:")
print(sample_df.columns.tolist())

# Descriptive statistics for the sample station
print(f"\nDescriptive statistics for sample station {sample_station_id}:")
display(sample_df.describe())

# Time series plot for altitude_diff for a few stations
plt.figure(figsize=(15, 8))
df = aquifer_by_stations[sample_station_id]
if 'altitude' in df.columns:
    plt.plot(df['date'], df['altitude'])

plt.title('Višina gladine podtalnice')
plt.xlabel('Datum')
plt.ylabel('Višina gladine')
plt.legend()
plt.grid(True)
plt.show()
```

## Surface waters

```python
watercourse_by_stations = joblib.load('../data/interim/watercourse_by_stations.joblib')
```

```python
watercourse_by_stations[1060].columns
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Basic statistics for each station
print(f"Total number of stations: {len(watercourse_by_stations)}")

# Create a summary dataframe
summary_stats = []

for station_id, df in watercourse_by_stations.items():
    stats = {
        'station_id': station_id,
        'num_records': len(df),
        'start_date': df['date'].min(),
        'end_date': df['date'].max(),
        'date_range_days': (df['date'].max() - df['date'].min()).days,
        'missing_values_pct': df.isna().mean().mean() * 100,
        'altitude_diff_mean': df['level_diff'].mean() if 'level_diff' in df.columns else np.nan,
        'altitude_diff_std': df['level_diff'].std() if 'level_diff' in df.columns else np.nan
    }
    summary_stats.append(stats)

summary_df = pd.DataFrame(summary_stats)
display(summary_df)

# Get a sample station to examine its columns
sample_station_id = list(watercourse_by_stations.keys())[0]
sample_df = watercourse_by_stations[sample_station_id]
print(f"\nColumns in sample station {sample_station_id}:")
print(sample_df.columns.tolist())

# Descriptive statistics for the sample station
print(f"\nDescriptive statistics for sample station {sample_station_id}:")
display(sample_df.describe())

# Time series plot for altitude_diff for a few stations
plt.figure(figsize=(15, 8))
df = watercourse_by_stations[sample_station_id]
if 'level_diff' in df.columns:
    plt.plot(df['date'], df['level_diff'])

plt.title('Level Difference Over Time')
plt.xlabel('Date')
plt.ylabel('Level Difference')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(15, 8))
if 'level' in df.columns:
    plt.plot(df['date'], df['level'])

plt.title('Višina gladine vode')
plt.xlabel('Datum')
plt.ylabel('Višina gladine')
plt.legend()
plt.grid(True)
plt.show()
```
